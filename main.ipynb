```python
import xmltodict
with open("DANFEBrota.xml", "rb") as arquivo:
    documento = xmltodict.parse(arquivo)
print(documento)
```

```python
dic_nota_fiscal = documento['nfeProc']['NFe']['infNFe']
valor_total = dic_nota_fiscal['total']['ICMSTot']['vNF']
CNPJ_vendedor = dic_nota_fiscal['emit']['CNPJ']
nome_vendedor = dic_nota_fiscal['emit']['xNome']
print(f"Valor total da nota fiscal:R$ {valor_total}", f"CNPJ do vendedor: {CNPJ_vendedor}", f"Nome do vendedor: {nome_vendedor}")
```

```python
CPF_comprador = dic_nota_fiscal['dest']['CPF']
nome_comprador = dic_nota_fiscal['dest']['xNome']
print(f"CPF do comprador: {CPF_comprador}", f"Nome do comprador: {nome_comprador}")
```

```python
produtos = dic_nota_fiscal['det']
lista = []
for produto in produtos:
    nome_produto = produto['prod']['xProd']
    valor_produto = produto['prod']['vProd']
    lista.append((nome_produto, valor_produto))
    print(f"Nome do produto: {nome_produto}", f"Valor do produto: R$ {valor_produto}")
```

```python
resposta = {
    "valor_total": [valor_total],
    "CNPJ_vendedor": [CNPJ_vendedor],
    "nome_vendedor": [nome_vendedor],
    "CPF_comprador": [CPF_comprador],
    "nome_comprador": [nome_comprador],
    "lista": [lista]
}
resposta
print(resposta)
```

```python
import pandas as pd

tabela = pd.DataFrame.from_dict(
    {
        "nome_produto": [produto[0] for produto in lista],
        "valor_produto": [produto[1] for produto in lista],
        "CNPJ_vendedor": [CNPJ_vendedor] * len(lista),
        "nome_vendedor": [nome_vendedor] * len(lista),
        "CPF_comprador": [CPF_comprador] * len(lista),
        "nome_comprador": [nome_comprador] * len(lista),
    }
)
tabela.to_excel("tabela_produtos.xlsx", index=False)
print(tabela)
```

```python
import os

pasta_xml = "."

listas_arquivos = os.listdir(pasta_xml)

for nome_arquivo in listas_arquivos:
    if nome_arquivo.endswith(".xml"):
        caminho_arquivo = os.path.join(pasta_xml, nome_arquivo)

        with open(caminho_arquivo, "rb") as arquivo_xml:
            documento = xmltodict.parse(arquivo_xml)

        if "nfeProc" not in documento:
            print(f"Arquivo ignorado: {nome_arquivo} não é uma NF-e.")
            continue

        dic_nota_fiscal = documento["nfeProc"]["NFe"]["infNFe"]
        valor_total = dic_nota_fiscal["total"]["ICMSTot"]["vNF"]
        CNPJ_vendedor = dic_nota_fiscal["emit"]["CNPJ"]
        nome_vendedor = dic_nota_fiscal["emit"]["xNome"]
        CPF_comprador = dic_nota_fiscal["dest"].get("CPF")
        nome_comprador = dic_nota_fiscal["dest"]["xNome"]
        produtos = dic_nota_fiscal["det"]

        if isinstance(produtos, dict):
            produtos = [produtos]

        lista = []
        for produto in produtos:
            nome_produto = produto["prod"]["xProd"]
            valor_produto = produto["prod"]["vProd"]
            lista.append((nome_produto, valor_produto))

        resposta = {
            "valor_total": [valor_total],
            "CNPJ_vendedor": [CNPJ_vendedor],
            "nome_vendedor": [nome_vendedor],
            "CPF_comprador": [CPF_comprador],
            "nome_comprador": [nome_comprador],
            "lista": [lista],
        }

        print(resposta)
```

```python
with open("NotaCariocaHashtag.xml", "rb") as arquivo:
    documento = xmltodict.parse(arquivo)

# XML da nota carioca é NFSe (não NF-e)
dic_nota_fiscal = (
    documento.get("ConsultarNfseResposta", {})
    .get("ListaNfse", {})
    .get("CompNfse", {})
    .get("Nfse", {})
    .get("InfNfse", {})
)

if not dic_nota_fiscal:
    raise KeyError("Estrutura da NFSe não encontrada em 'ConsultarNfseResposta'.")

# Campos da NFSe
servico = dic_nota_fiscal.get("Servico", {})
valores = servico.get("Valores", {})
valor_total = valores.get("ValorServicos")

prestador = dic_nota_fiscal.get("PrestadorServico", {})
CNPJ_vendedor = (
    prestador.get("IdentificacaoPrestador", {}).get("Cnpj")
    or prestador.get("Cnpj")
)
nome_vendedor = prestador.get("RazaoSocial")

tomador = dic_nota_fiscal.get("TomadorServico", {})
identificacao_tomador = tomador.get("IdentificacaoTomador", {})
CPF_comprador = identificacao_tomador.get("Cpf") or tomador.get("Cpf")
nome_comprador = tomador.get("RazaoSocial")

print(
    f"Valor total da nota fiscal: R$ {valor_total}",
    f"CNPJ do vendedor: {CNPJ_vendedor}",
    f"Nome do vendedor: {nome_vendedor}",
    f"CPF do comprador: {CPF_comprador}",
    f"Nome do comprador: {nome_comprador}",
)
```
